```python
entero = 42
decimal = 3.14
texto = "100"
logico = True

print("Tipos:", type(entero), type(decimal), type(texto), type(logico))

# Conversiones
print("str -> int:", int(texto) + 8)        # "100" -> 100
print("int -> float:", float(entero))
print("int -> str:", "edad: " + str(entero)) # hay que convertir para concatenar
print("int -> bool:", bool(0), bool(42))     # 0 es False, el resto True
```
